````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# 1. убираем пробелы и приводим к виду Almaty
cities_clean = []
for c in cities:
    cities_clean.append(c.strip().title())

print(cities_clean)
````

````python
# 2. суммы из строк в int
amounts_int = []
for a in amounts:
    amounts_int.append(int(a))

print(amounts_int)
````

````python
# 3. кол-во заказов, сумма, средний чек
n_orders = len(amounts_int)
total = sum(amounts_int)
avg_check = round(total / n_orders)

print(f"Заказов всего: {n_orders}")
print(f"Общая сумма:   {total} ₸")
print(f"Средний чек:   {avg_check} ₸")
````

````python
# 4. неоплаченные заказы
n_unpaid = 0
unpaid_sum = 0
for a, p in zip(amounts_int, paid):
    if not p:
        n_unpaid += 1
        unpaid_sum += a

print(f"Не оплачено: {n_unpaid} заказа на сумму {unpaid_sum} ₸")
````

````python
# 5. таблица город | сумма | статус
print(f"{'город':<10} | {'сумма':>6} | статус")
print("-" * 34)
for c, a, p in zip(cities_clean, amounts_int, paid):
    status = "оплачен" if p else "НЕ оплачен"
    print(f"{c:<10} | {a:>6} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# 6. выручка по Алматы и Астане (только оплаченные)
rev_almaty = 0
rev_astana = 0
for c, a, p in zip(cities_clean, amounts_int, paid):
    if not p:
        continue
    if c == "Almaty":
        rev_almaty += a
    elif c == "Astana":
        rev_astana += a

print(f"Выручка Almaty: {rev_almaty} ₸")
print(f"Выручка Astana: {rev_astana} ₸")
````

````python
# 7. самый крупный оплаченный заказ
max_idx = -1
for i in range(len(amounts_int)):
    if paid[i] and (max_idx == -1 or amounts_int[i] > amounts_int[max_idx]):
        max_idx = i

print(f"Индекс: {max_idx}, город: {cities_clean[max_idx]}, сумма: {amounts_int[max_idx]} ₸")
````

````python
# 8. доля заказов в сб и вс
weekend = 0
for d in days:
    if d == "сб" or d == "вс":
        weekend += 1

weekend_share = weekend / len(days) * 100
print(f"Заказов в выходные: {weekend} из {len(days)} → {weekend_share:.1f}%")
````

````python
# 9. копилка на курс за 400 000
price = 400_000

# по 15 000 каждый месяц
saved = 0
months = 0
while saved < price:
    saved += 15_000
    months += 1
print(f"По 15 000 ₸ в месяц: {months} мес. (накоплено {saved} ₸)")

# каждый месяц на 1 000 больше
saved = 0
months = 0
monthly = 15_000
while saved < price:
    saved += monthly
    months += 1
    monthly += 1_000
print(f"С ростом на 1 000 ₸:  {months} мес. (накоплено {saved} ₸)")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# 10. выручка по городам без словаря: два списка, индексы совпадают
unique_cities = []
revenue_by_city = []
for c, a, p in zip(cities_clean, amounts_int, paid):
    if c not in unique_cities:
        unique_cities.append(c)
        revenue_by_city.append(0)
    if p:
        revenue_by_city[unique_cities.index(c)] += a

# сортировка выбором по убыванию, меняем местами сразу в обоих списках
for i in range(len(revenue_by_city)):
    best = i
    for j in range(i + 1, len(revenue_by_city)):
        if revenue_by_city[j] > revenue_by_city[best]:
            best = j
    revenue_by_city[i], revenue_by_city[best] = revenue_by_city[best], revenue_by_city[i]
    unique_cities[i], unique_cities[best] = unique_cities[best], unique_cities[i]

print(unique_cities)
print(revenue_by_city)
print()
for c, r in zip(unique_cities, revenue_by_city):
    print(f"{c:<10} {r:>6} ₸")
````

````python
# 11. медиана без statistics и numpy
s = sorted(amounts_int)
n = len(s)
mid = n // 2
if n % 2 == 1:
    median = s[mid]  # нечётное кол-во
else:
    median = (s[mid - 1] + s[mid]) / 2  # чётное: среднее двух средних

print(f"Отсортированные суммы: {s}")
print(f"Медиана: {median} ₸  (для сравнения, среднее: {avg_check} ₸)")
````

````python
# 12. гистограмма выручки по дням, один блок = 5000
day_names = []
day_revenue = []
for d, a, p in zip(days, amounts_int, paid):
    if d not in day_names:
        day_names.append(d)
        day_revenue.append(0)
    if p:
        day_revenue[day_names.index(d)] += a

step = 5_000
width = max(day_revenue) // step  # чтобы числа стояли ровно
for d, r in zip(day_names, day_revenue):
    bar = "█" * (r // step)
    print(f"{d} | {bar:<{width}}  {r}")
````
