````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

````python
# 2. оплаченные и неоплаченные через маску
paid_sum = amount[paid].sum()
unpaid_sum = amount[~paid].sum()
print(f"оплаченных заказов: {paid.sum()}, сумма {paid_sum:,.0f} ₸")
print(f"неоплаченных заказов: {(~paid).sum()}, сумма {unpaid_sum:,.0f} ₸")

# сколько оплаченной суммы приходится на TEST
no_test = cust != "TEST"
print(f"из оплаченной суммы на TEST: {amount[paid & ~no_test].sum():,.0f} ₸")
````

````python
# 4. скидка 15% на оплаченные заказы дороже 50 000
big = paid & (amount > 50_000)
loss_all = (amount[big] * 0.15).sum()
loss_no_test = (amount[big & no_test] * 0.15).sum()

print(f"со всеми заказами: {loss_all:,.0f} ₸ (заказов: {big.sum()})")
print(f"без TEST: {loss_no_test:,.0f} ₸ (заказов: {(big & no_test).sum()})")
print(f"разница в {loss_all / loss_no_test:.1f} раза")
````

````python
# 5. доля заказов, доставленных за 1-2 дня, по городам
# заказы без даты доставки (nan) не считаем, меньше 1 дня в данных нет
has_days = ~np.isnan(days)
for c in np.unique(city):
    m = (city == c) & has_days
    fast = (days[m] <= 2).mean()
    print(f"{c:<10} {fast:.1%}  (заказов с доставкой: {m.sum()})")
````

## [Челлендж]
7. Сравните скорость: сумма оплаченных заказов циклом `for` и через маску. Во сколько раз быстрее?
8. Смоделируйте A/B-тест. `np.random.seed(42)`. Группа A: 1 000 клиентов, вероятность покупки 5%; группа B: 1 000 клиентов, 6%
   (покупка = `np.random.rand(1000) < 0.05`). Сколько покупок в каждой группе?
   Повторите эксперимент 100 раз: в скольких случаях B оказалась лучше A? Что это говорит о размере выборки? (2–3 предложения)

````python
# 7. цикл против маски
import timeit

def loop_sum():
    s = 0
    for a, p in zip(amount, paid):
        if p:
            s += a
    return s

def mask_sum():
    return amount[paid].sum()

print(f"суммы совпадают: {loop_sum() == mask_sum()}, {mask_sum():,.0f} ₸")

# берём лучший из 5 замеров, так меньше влияет фоновая нагрузка
t_loop = min(timeit.repeat(loop_sum, number=20, repeat=5)) / 20
t_mask = min(timeit.repeat(mask_sum, number=2000, repeat=5)) / 2000
print(f"цикл: {t_loop * 1000:.2f} мс, маска: {t_mask * 1000:.3f} мс")
print(f"во сколько раз маска быстрее: {t_loop / t_mask:.0f}")
````

````python
# 8. A/B-тест
np.random.seed(42)
a_buy = np.random.rand(1000) < 0.05
b_buy = np.random.rand(1000) < 0.06
print("покупок в A:", a_buy.sum(), "  в B:", b_buy.sum())

# повторяем эксперимент 100 раз
b_better = 0
for _ in range(100):
    a_cnt = (np.random.rand(1000) < 0.05).sum()
    b_cnt = (np.random.rand(1000) < 0.06).sum()
    if b_cnt > a_cnt:
        b_better += 1
print("B лучше A в", b_better, "экспериментах из 100")
````
